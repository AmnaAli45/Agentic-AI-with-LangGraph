```python
from langgraph.store.memory import InMemoryStore
```

```python
# Creating Store 
store = InMemoryStore()
```

```python
# Creating the namespace 
namespace = ("users","u1")
```

# Creating Memory

```python
store.put(namespace,'1',{"data": "user likes pizza."})
store.put(namespace,'2',{"data": "user prefers dark mode."})
```

```python
# another namespace
namespace2 = ("users","u2")

store.put(namespace,'1',{"data": "user likes pasta."})
store.put(namespace,'2',{"data": "user prefers grid style navigation."})
```

# Retreiving Memory 

```python
store.get(namespace,'1')
```

# Retreiving all Memory

```python
items = store.search(namespace)

for item in items:
    print(item.value)
```

# Semantic Search 

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding_model = OpenAIEmbeddings(model="text-embedding-3-small", dimensions=1024)
```

```python
store = InMemoryStore(embedding_model=embedding_model)
```

```python
namespace3 = ("users","u3")
```

```python
store.put(namespace,'1',{"data": "user likes pasta."})
store.put(namespace,'2',{"data": "user prefers grid style navigation."})
store.put(namespace,'3',{"data": "user likes pizza."})
store.put(namespace,'4',{"data": "user likes dark theme."})
store.put(namespace,'5',{"data": "user is very intelligent."})
store.put(namespace,'6',{"data": "user often visits the website."})
store.put(namespace,'7',{"data": "user develops the application in flutter."})
store.put(namespace,'8',{"data": "user gives me appreciation."})
```

```python
items = store.search(namespace3, query="what does the user like?",limit = 3) # --> ye limit decide kre ga ki kitne similar items chahiye humein

for item in items:
    print(item.value)
```
